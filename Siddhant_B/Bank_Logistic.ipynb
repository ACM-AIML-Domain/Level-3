## Sigmoid Function

```python
def sigmoid(z):
    return 1/(1+np.exp(-z))
```

## Propagation Function

```python
def propagation(x,y,w,b):

    
    
    m = x.shape[0]
    z = np.dot(x,w) + b
    y_hat = sigmoid(z)

    print(x.shape)

    epsilon = 1e-15
    cost = (-1/m) * np.sum(y*np.log(y_hat + epsilon) + (1-y)*np.log(1-y_hat + epsilon))

    dw = 1/m * np.dot(x.T, (y_hat-y))
    db = 1/m * np.sum(y_hat-y)

    return cost, dw, db

```

## Gradient Descent

```python
def gradient(w,b,alpha,dw,db):

    w = w - alpha * dw
    b = b - alpha* db

    print(w.shape)
    
    return w,b  

```

```python
def logistic_reg(x,y):
      
    w = np.zeros(x.shape[1])

    b = 0
    
    alpha = 0.01

    iterations = 1000

    y_hat = np.zeros(y.shape[0])

    for i in range(iterations):

        cost, dw, db = propagation(x, y, w, b)
        w,b = gradient(w,b,dw,db,alpha)

        if i % 100 == 0:
            print(i,":",cost, "\n")

    return w,b
```

```python
y_pred = np.zeros(y_test.shape[0])
```

```python
def final_test(x_train,y_train, x_test, y_test):
    w,b = logistic_reg(x_train, y_train)
    z = np.dot(x_test, w) + b
    y_pred = sigmoid(z)
    return y_pred

```

```python
y_pred = final_test(x_train, y_train, x_test, y_test)
y_pred_class = (y_pred >= 0.5).astype(int)
accuracy = np.mean(y_pred_class == y_test)
print(f"Test Accuracy: {accuracy:.4f}")
```
