```python
import pandas as pd
import numpy as np
```

```python
try:
    df = pd.read_csv("Boston.csv")
    print("Dataset loaded successfully.")
except FileNotFoundError:
    print("Error: Boston.csv not found. Please ensure the file is in the correct directory.")
    exit()
```

```python
if 'Unnamed: 0' in df.columns:
    df = df.drop('Unnamed: 0', axis=1)
```

```python
Y = df['medv'].values.reshape(-1, 1)
X_raw = df[['lstat', 'rm']].values

X = np.insert(X_raw, 0, 1, axis=1)

XT = X.T

XTX = XT.dot(X)

XTX_inv = np.linalg.inv(XTX)

XTY = XT.dot(Y)
```

```python
beta_coefficients = XTX_inv.dot(XTY)

intercept = beta_coefficients[0][0]
lstat_coeff = beta_coefficients[1][0]
rm_coeff = beta_coefficients[2][0]

Y_hat = X.dot(beta_coefficients)

residuals = Y - Y_hat

SST = np.sum((Y - np.mean(Y))**2)

SSR = np.sum(residuals**2)

R_squared = 1 - (SSR / SST)
```

```python
R_squared = 1 - (SSR / SST)

print("--- Multiple Linear Regression Model Results (Numpy Only) ---")
print("Target: medv | Predictors: lstat, rm\n")

print("Calculated Coefficients (beta):")
print(f"Intercept (B0): {intercept:.3f}")
print(f"lstat Coefficient (B1): {lstat_coeff:.3f}")
print(f"rm Coefficient (B2): {rm_coeff:.3f}")

print("\n--- Model Evaluation ---")
print(f"R-squared: {R_squared:.3f}")

print("\n--- Interpretation ---")
print(f"The model explains {R_squared*100:.1f}% of the variance in Median Home Value (medv).")
print(f"Holding 'lstat' constant, a 1 unit increase in 'rm' is associated with an increase of ${rm_coeff*1000:.0f} in home value.")
print(f"Holding 'rm' constant, a 1 unit increase in 'lstat' is associated with a decrease of ${abs(lstat_coeff*1000):.0f} in home value.")
```
